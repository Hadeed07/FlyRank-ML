####Freestyle Lane Predicting the trend of content
I want to see how the prediction model works and how it is highly linked to data engineering.

The problem is about the prediction of the already existing content and their forecaste momentum.
It is predicting whether the existing piece of content is losing or gaining the search distribution momentum.

This lane offers the insights why a piece of content would go up or down and what are the key features for it.
For example. If a particular content is gaining impression but its click through ratio goes down from 5% to 2%, it is worth noting.
Similarly, an article can have increasing impressions but the average position is fall from 5 to 12, which suggests the search demand is still there, competition is still there but the content is losing its ranking.

Broadly, it is like asking, which type of content will exhibits the accelerating search momentum while maintaining its ability to capture traffic.

This means decline is already the dominant pattern in the data — decline detection alone would have wide applicability across most clients, which is why an early-warning model is worth building rather than treating each dip as a one-off.

Based on these correlated features, an ML model can be designed to predict the trend. It serves as an early-warning system, so human investigation happens at the proper time — before the signal fully matures.

## 2. The question: decision, action, cost of a wrong call


###What decision does your work improve?
Whether a content team should proactively intervene on a piece of content — refresh it, promote it, or leave it alone — before its performance actually moves. Without your model, they only find out a page declined after it's already declined (reactive).

###Who acts on it?
Realistically, a content or SEO team member reviewing a prioritized list — not an automated system. This matters a lot for scope: your output isn't "the model decides," it's "the model ranks candidates for a human to check."


###What does a wrong recommendation cost?
Two different costs, and they're asymmetric:\
False positive (flagged as declining, but it wasn't going to): wasted review time — someone checks a page that didn't need it. Annoying, low-cost.
False negative (a page silently declines and your model missed it): the team never sees it, so nothing gets fixed. This is the more expensive failure, since it means the model failed at its actual job with no fallback.

`impressions_90d` ≈ `impressions_prev_30d` + `impressions_last_30d` + some remainder — it is not an independent window, it's a rollup that includes the most recent 30 days. Since `impressions_last_30d` overlaps with the window the target (`trend_pct`/`trend_direction`) is computed from, any `_90d` column would leak future information into the features. Conclusion: exclude all `_90d` columns from the model; use only `prev_30d` and static content attributes.

```python
df[['ctr', 'avg_position', 'engagement_rate', 'scroll_rate', 'ai_traffic_pct', 'trend_pct']].corr()
```

`ctr`, `avg_position`, `engagement_rate`, `scroll_rate`, and `ai_traffic_pct` all show weak correlation with `trend_pct` (close to 0). This is expected — these are current-state ratios, not measures of directional change, so a weak standalone correlation doesn't rule them out as useful features; they may still combine with other signals non-linearly (part of why a rule-based cutoff won't work and ML is justified).

Open question: these columns' exact source window (does `ctr` reflect prev_30d or last_30d?) is unconfirmed — needs resolution before they're safely added to the feature set.

Conclusion: the split must be grouped by `client_id`, not random, so no client's pages appear in both sets. This is only one slice of class imbalance.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

Observed:<br> "Content with [X characteristic] showed [Y trend] over the observed 30/90-day window." These are descriptions of what happened in this dataset, not universal laws.


Directional:<br> "Pages with lower freshness scores and higher content age were disproportionately represented among decliners." This tells you the direction of association, and roughly how strong it is — not a guarantee for any individual page.


Decision-support:<br> "This model ranks pages by estimated risk of decline, so a content team can prioritize review — it does not replace human judgment, it narrows where a human should look first." The output is a ranked shortlist for review, not an automated verdict.
